### Данные: американские банки

FDIC — американский аналог АСВ — публикует квартальную отчётность каждого банка
страны и список банкротств. Мы взяли оттуда 2001–2026 годы: все банки, которые
за это время упали, плюс случайную выборку выживших.

Денежные величины — **в тысячах долларов** (соглашение FDIC).

```python
panel = pd.read_parquet(DATA + "banks_panel.parquet")
print(f"строк (банк-квартал): {len(panel):,}")
print(f"банков: {panel.cert.nunique():,}   кварталы: "
      f"{panel.repdte.min().date()} .. {panel.repdte.max().date()}")
panel[["cert", "name", "state", "repdte", "assets", "equity",
       "loans_gross", "npl", "net_income", "fail_date"]].head()
```

Одна строка — это **один банк в один квартал**. `cert` — номер лицензии FDIC,
он и есть ключ банка (имена меняются, номер нет). `fail_date` заполнена только у
тех, кто впоследствии рухнул.

Работать с сырыми суммами неудобно: банк на 300 миллионов и банк на 300
миллиардов несравнимы. Поэтому в таблице уже посчитаны **ратио** — суммы,
поделённые на масштаб банка. Это стандартная надзорная схема CAMELS:

| Префикс | Что меряет | Пример |
|---|---|---|
| `c_` | **C**apital — капитал | `c_equity_assets` = капитал / активы |
| `a_` | **A**sset quality — качество активов | `a_npl_gross_loans` = проблемные кредиты / кредиты |
| `m_` | **M**anagement — эффективность | `m_cost_income` = расходы / процентный доход |
| `e_` | **E**arnings — прибыльность | `e_roa` = прибыль / активы |
| `l_` | **L**iquidity — ликвидность | `l_loans_deposits` = кредиты / депозиты |
| `s_` | **S**ensitivity — рыночный риск | `s_securities_assets` = ценные бумаги / активы |

### Настоящие данные грязные

Прежде чем что-то считать — обязательный шаг, которого нет в учебниках.
В реальной отчётности встречаются банки с нулевыми депозитами, отрицательным
капиталом, процентным доходом около нуля (у которого `m_cost_income` улетает в
тысячи). Это не ошибки FDIC — это экзотические учреждения: трастовые компании,
карточные банки, банки в процессе ликвидации.

Отфильтруем их **явно**, чтобы решение было видно и его можно было оспорить.

```python
banks = panel[
    (panel.assets > 50_000)                       # больше 50 млн $ — не микроструктуры
    & (panel.loans_gross > 0) & (panel.deposits > 0) & (panel.nii > 0)
    & panel.c_equity_assets.between(0, 0.5)       # капитал от 0 до 50% активов
    & panel.a_npl_gross_loans.between(0, 0.5)
    & panel.m_cost_income.between(0, 3)
    & panel.l_loans_deposits.between(0, 2)
    & panel.a_llr_npl_coverage.between(0, 10)
    & panel.e_nim.between(0, 0.12)                # маржа выше 12% — не банк
    & panel.a_nco_rate.between(-0.02, 0.15)
    & panel.e_roa.between(-0.15, 0.10)            # ROA от -15% до +10% годовых
].copy()
print(f"было {len(panel):,} строк  ->  осталось {len(banks):,} "
      f"({len(banks)/len(panel):.0%}), банков {banks.cert.nunique():,}")
print(f"из них перед крахом (в ближайшие 4 квартала): "
      f"{int(banks.failed_next_4q.sum()):,} = {banks.failed_next_4q.mean():.2%}")
```

Треть строк отброшена. **Это не безобидно**: мы только что изменили популяцию, о
которой будем делать выводы. Все дальнейшие утверждения относятся к «обычным
работающим банкам», а не ко всем учреждениям с лицензией FDIC. Проговаривать
такие решения вслух — часть профессии; молча их делать — способ получить
неприятный вопрос на защите.

### Словарь

* **Отклик** (response), $Y$ — то, что хотим узнать. Возьмём `e_roa`,
  рентабельность активов: сколько прибыли банк приносит на доллар активов.
* **Признаки** (predictors), $X_1, \dots, X_p$ — то, через что собираемся узнать.
  Ратио CAMELS.
* **Наблюдение** — одна строка, один банк в один квартал.

Предполагается, что связь есть, и записывается она так:

$$ Y = f(X) + \varepsilon $$

**$f$** — фиксированная, но неизвестная функция: *систематическая* часть того,
что $X$ сообщает про $Y$. **$\varepsilon$** — ошибка: всё остальное, в среднем
ноль и независимо от $X$.

Дальше вся статистика — способы оценить $f$ по данным.

Посмотрим на три ратио по отдельности. Точек сто тысяч, поэтому рисуем случайную
подвыборку, а поверх — прямую наименьших квадратов по **всем** данным.

```python
fig = px.scatter(banks, 
    x="repdte", 
    y="l_liquid_assets_ratio", 
    color="failed_next_4q",
    trendline="lowess",
    opacity=0.2
    )

fig.update_traces(marker=dict(opacity=0.01), selector=dict(mode='markers'))
fig.show()
```

```python
PAIRS = [("a_nco_rate", "Списания / кредиты"),
         ("m_cost_income", "Расходы / процентный доход"),
         ("e_nim", "Процентная маржа")]
show = banks.sample(4000, random_state=SEED)

fig = make_subplots(rows=1, cols=3, shared_yaxes=True,
                    subplot_titles=[t for _, t in PAIRS])
for j, (v, _) in enumerate(PAIRS, start=1):
    b, a = np.polyfit(banks[v], banks["e_roa"], 1)      # наклон и свободный член
    grid = np.linspace(banks[v].quantile(.001), banks[v].quantile(.999), 50)
    fig.add_scatter(x=show[v], y=show["e_roa"], mode="markers", row=1, col=j,
                    marker=dict(color=ORANGE, size=3, opacity=0.35), showlegend=False)
    fig.add_scatter(x=grid, y=a + b * grid, mode="lines", row=1, col=j,
                    line=dict(color=BLUE, width=4), showlegend=False)
    fig.update_xaxes(title_text=v, row=1, col=j)
fig.update_yaxes(title_text="ROA (годовой)", row=1, col=1, range=[-0.06, 0.04])
fig.update_layout(title="Рентабельность против трёх ратио CAMELS", **LAYOUT)
fig.show()
```

Три вывода, которые видно глазами.

1. **Связь есть везде**, и знаки осмысленные: списанные кредиты и расходы тянут
   прибыльность вниз, процентная маржа — вверх.
2. **Сила разная.** Списания в одиночку объясняют 31% разброса ROA, расходы —
   26%, маржа — 12%. Это и есть «сколько $X$ знает про $Y$», и различается оно в
   разы.
3. **Разброс вокруг прямой — это $\varepsilon$.** Банк выше прямой заработал
   больше, чем объясняется его ратио. Почему? Регион, отрасль заёмщиков, качество
   менеджмента, удача. Ничего этого в таблице нет — значит, для модели это шум.

Проверим числами: доля объяснённой дисперсии ($R^2$) у каждой пары.

```python
import statsmodels.formula.api as smf

for v, label in PAIRS:
    m = smf.ols(f"e_roa ~ {v}", data=banks).fit()
    print(f"{label:32s} R^2 = {m.rsquared:.3f}   наклон = {m.params[v]:+8.4f}"
          f"   t = {m.tvalues[v]:+7.1f}")
```

## 2 · Зачем нужна $f$: предсказание и вывод

Оценив $\hat f$, мы можем делать две принципиально разные вещи.

### Предсказание

Признаки $X$ доступны легко, отклик $Y$ — нет. Считаем $\hat Y = \hat f(X)$, и
**форма $\hat f$ нас не волнует**: пусть будет чёрным ящиком, лишь бы $\hat Y$
был близок к $Y$.
*Вопрос надзорщика:* «вот отчётность банка за квартал — какова вероятность, что
он упадёт в течение года?»

### Вывод (inference)

Здесь интересна как раз **форма** $f$: какие признаки связаны с откликом, с
каким знаком, насколько сильно, можно ли этому верить. Чёрный ящик не годится.
*Вопрос регулятора:* «если поднять требования к капиталу на процентный пункт,
что будет с прибыльностью сектора?»

Разница не академическая — сейчас увидим, как одни и те же данные дают на эти
два вопроса **противоположные ответы**.

### Неснижаемая ошибка

Сначала про предел точности. Возьмём 40 случайных банков одного квартала,
проведём через них гладкую кривую (это наша $\hat f$) и соединим отрезком
каждый банк с кривой.

```python
snap = (banks[banks.repdte == "2015-12-31"]
        .sample(40, random_state=SEED).sort_values("m_cost_income"))
sm = np.poly1d(np.polyfit(snap.m_cost_income, snap.e_roa, 3))   # гладкая f-с-крышкой
grid = np.linspace(snap.m_cost_income.min(), snap.m_cost_income.max(), 100)

fig = go.Figure()
for _, r in snap.iterrows():                                    # отрезки = ошибки
    fig.add_scatter(x=[r.m_cost_income] * 2, y=[r.e_roa, sm(r.m_cost_income)],
                    mode="lines", line=dict(color=GREY, width=1), showlegend=False)
fig.add_scatter(x=snap.m_cost_income, y=snap.e_roa, mode="markers", name="банки",
                marker=dict(color=ORANGE, size=9))
fig.add_scatter(x=grid, y=sm(grid), mode="lines", name="оценка f",
                line=dict(color=BLUE, width=4))
fig.update_layout(title="Каждый отрезок — это ε: то, чего ратио не объясняют",
                  xaxis_title="Расходы / процентный доход",
                  yaxis_title="ROA (годовой)", **LAYOUT)
fig.show()
```

Отрезки не исчезнут, как бы мы ни улучшали кривую. У двух банков с одинаковым
соотношением расходов к доходу прибыльность разная — потому что различаются
регион, состав заёмщиков, ставки по вкладам, качество кредитного комитета.
В таблице этого нет. Значит, для модели это $\varepsilon$.

Отсюда — центральное для всего курса разложение. Для новой точки $X$:

$$
\underbrace{\mathbb{E}\big[(Y - \hat{Y})^2\big]}_{\text{ожидаемая ошибка}}
= \underbrace{\big[f(X) - \hat f(X)\big]^2}_{\text{снижаемая}}
+ \underbrace{\operatorname{Var}(\varepsilon)}_{\text{неснижаемая}}
$$

**Снижаемая** часть — наша работа: лучше модель, больше данных, умнее признаки.
**Неснижаемая** не убирается **ничем**: даже подари нам кто-нибудь истинную $f$,
ошибка осталась бы.

Практический смысл: у точности есть пол, и он неизвестен. Обещать заказчику
точность ниже этого пола — обещать невозможное. Проверим на симуляции, где мы
сами задаём $f$ и $\sigma$, а потому знаем, где пол.

```python
def f_true(x):
    return np.sin(1.5 * x) + 0.5 * x      # истина, которую в жизни никто не показывает
sigma = 0.40                              # sd шума  ->  неснижаемая ошибка = sigma**2

x_tr = rng.uniform(-3, 3, 80);   y_tr = f_true(x_tr) + rng.normal(0, sigma, 80)
x_te = rng.uniform(-3, 3, 3000); y_te = f_true(x_te) + rng.normal(0, sigma, 3000)
degrees = list(range(0, 11))
test_mse = [np.mean((np.polyval(np.polyfit(x_tr, y_tr, d), x_te) - y_te) ** 2)
            for d in degrees]

fig = go.Figure()
fig.add_scatter(x=degrees, y=test_mse, mode="lines+markers", name="ошибка на новых данных",
                line=dict(color=BLUE, width=3), marker=dict(size=9))
fig.add_hline(y=sigma ** 2, line=dict(color=ORANGE, width=3, dash="dash"),
              annotation_text=f"пол: Var(ε) = {sigma**2:.2f}", annotation_position="top right")
fig.update_layout(title="Ошибка падает до пола и не пробивает его",
                  xaxis_title="Гибкость модели (степень полинома)",
                  yaxis_title="MSE на новых данных", **LAYOUT)
fig.show()
print(f"лучшая достигнутая ошибка = {min(test_mse):.3f}   |   пол Var(ε) = {sigma**2:.3f}")
```

### Предсказание и вывод дают разные ответы

Теперь обещанная демонстрация. Вопрос содержательный:

> **Помогает ли прибыльности щедрое резервирование?**

`a_llr_npl_coverage` — покрытие проблемных кредитов резервами: сколько рублей
резерва приходится на рубль плохого кредита. Аккуратный банк держит его высоким.

```python
simple = smf.ols("e_roa ~ a_llr_npl_coverage", data=banks).fit()
b_s = simple.params["a_llr_npl_coverage"]
print(simple.summary().tables[1])
print(f"R^2 = {simple.rsquared:.3f}   n = {int(simple.nobs):,}")
print(f"единица покрытия -> {b_s * 1e4:+.0f} базисных пунктов ROA")
```

$t = 63$, $p$ неотличимо от нуля, каждая единица покрытия — плюс 14 базисных
пунктов ROA. Хочется сделать вывод: *резервируйте щедрее, это окупается.*

Теперь тот же вопрос, но в модели, где есть остальные очевидные факторы
прибыльности. Это и есть **полная спецификация**, к которой мы будем
возвращаться весь ноутбук.

```python
FEATURES = ["e_nim", "m_cost_income", "a_npl_gross_loans", "a_nco_rate",
            "c_equity_assets", "l_loans_deposits", "s_securities_assets",
            "l_brokered_share", "a_cre_concentration", "e_earnings_vol_8q"]
model_df = banks.dropna(subset=["e_roa", "a_llr_npl_coverage"] + FEATURES)
multi = smf.ols("e_roa ~ a_llr_npl_coverage + " + " + ".join(FEATURES),
                data=model_df).fit()
b_m = multi.params["a_llr_npl_coverage"]
print(multi.summary().tables[1])
print(f"R^2 = {multi.rsquared:.3f}   n = {int(multi.nobs):,}")
print(f"\nкоэффициент при покрытии: {b_s:.5f}  ->  {b_m:.6f}   "
      f"(в {b_s / b_m:.0f} раз меньше)")
```

Коэффициент упал **в 17 раз**: с 14 базисных пунктов до 0.8. Эффект, который в
парной регрессии выглядел содержательным, оказался почти целиком заимствованным.

Обратите внимание на тонкость, которая будет мучить вас весь курс: $t$ при
покрытии остался **выше 4**, то есть формально коэффициент по-прежнему «значим».
При 96 тысячах наблюдений значимо практически всё. **Значимость и величина —
разные вопросы**, и здесь величина говорит «забудьте», а значимость — «оставьте».
Слушать надо величину.

Откуда взялась заимствованная связь? Посмотрим на корреляции.

```python
print("корреляции покрытия с остальным:")
print(model_df[["a_llr_npl_coverage", "a_npl_gross_loans", "a_nco_rate",
                "e_roa"]].corr()["a_llr_npl_coverage"].round(3).to_string())
```

Разгадка: **корреляция покрытия с долей проблемных кредитов равна −0.41**.
Высокое покрытие бывает не у щедрых банков, а у тех, **у кого мало плохих
кредитов** — знаменатель маленький, отношение большое. А мало плохих кредитов —
это и есть причина высокой прибыли.

Парная регрессия честно измерила: «у банков с высоким покрытием прибыль выше».
Множественная измерила другое: «при том же качестве активов покрытие не даёт
почти ничего». Второе — ответ на вопрос, который мы задавали.

> **Что уносим.** Для **предсказания** годятся обе модели. Для **вывода** они
> противоположны, и решение «требовать ли от банков больше резервов ради
> прибыльности» зависит от того, какую вы прочли. Один датасет, два вопроса,
> разные ответы. Это стержень всего курса.

## 3 · Как оценивают $f$: параметрические и непараметрические модели

### Параметрические модели: два шага

**Шаг 1.** Предполагаем форму $f$. Например, линейную:

$$ f(X) = \beta_0 + \beta_1 X_1 + \dots + \beta_p X_p $$

**Шаг 2.** Подбираем $p+1$ чисел так, чтобы модель села на данные.

Задача «угадать функцию» свелась к задаче «подобрать несколько чисел». Это
огромное упрощение — и в этом же риск: если истинная $f$ не такая, никакой подбор
коэффициентов не спасёт.

Модель из §2 объясняет **55% разброса ROA** десятью числами. Для квартальной
прибыльности банка это много: половина того, что отличает прибыльный банк от
убыточного, читается прямо из отчётности.

Посмотрим на неё в трёх измерениях. Взять сразу десять осей нельзя, поэтому
зафиксируем восемь ратио на медиане и оставим два самых сильных: **списания**
(сколько кредитов признано безнадёжными) и **кредиты к депозитам** (насколько
банк «вложился» в кредитование).

```python
def surface3d(xg, yg, Z, labels, points=None, sticks=None, title="", height=560):
    """3-D поверхность над сеткой (xg, yg).

    Сетка задаётся ИНДЕКСАМИ, а настоящие значения выносятся в подписи осей:
    в некоторых сборках plotly поверхность не рисуется, если координаты сетки —
    произвольные числа. По индексам работает всегда.
    """
    ix = lambda v, g: (np.asarray(v, float) - g[0]) / (g[-1] - g[0]) * (len(g) - 1)
    fig = go.Figure(go.Surface(z=Z, colorscale="YlGnBu_r", opacity=0.85, showscale=False))
    if sticks is not None:
        for x, y, z0, z1 in sticks:
            fig.add_scatter3d(x=[ix(x, xg)] * 2, y=[ix(y, yg)] * 2, z=[z0, z1],
                              mode="lines", line=dict(color="#333333", width=3),
                              showlegend=False)
    if points is not None:
        px_, py_, pz_ = points
        fig.add_scatter3d(x=ix(px_, xg), y=ix(py_, yg), z=pz_, mode="markers",
                          marker=dict(color="red", size=4), showlegend=False)
    def axis(g, name):
        idx = np.linspace(0, len(g) - 1, 5)
        return dict(title=name, tickvals=idx,
                    ticktext=[f"{g[int(round(i))]:.3g}" for i in idx])
    fig.update_layout(title=title, height=height, margin=dict(l=0, r=0, t=45, b=0),
                      scene=dict(xaxis=axis(xg, labels[0]), yaxis=axis(yg, labels[1]),
                                 zaxis=dict(title=labels[2])))
    return fig
print("хелпер готов")
```

Красные точки — 40 случайных банков, чёрные отрезки — их ошибки: расстояние от
банка до плоскости.

```python
X1, X2 = "a_nco_rate", "l_loans_deposits"
LBL = ("Списания / кредиты", "Кредиты / депозиты", "ROA")
linear = smf.ols("e_roa ~ " + " + ".join(FEATURES), data=model_df).fit()

xg = np.linspace(model_df[X1].quantile(.01), model_df[X1].quantile(.99), 30)
yg = np.linspace(model_df[X2].quantile(.01), model_df[X2].quantile(.99), 30)
XG, YG = np.meshgrid(xg, yg)
med = model_df[FEATURES].median()

def partial_surface(model, extra=None):
    """Предсказание модели на сетке, остальные ратио — на медиане."""
    scen = pd.DataFrame({c: np.full(XG.size, med[c]) for c in FEATURES})
    scen[X1], scen[X2] = XG.ravel(), YG.ravel()
    return model.predict(scen).to_numpy().reshape(XG.shape)

S = model_df.sample(40, random_state=SEED)
fit40 = linear.predict(S)
surface3d(xg, yg, partial_surface(linear), LBL,
          points=(S[X1], S[X2], S.e_roa),
          sticks=list(zip(S[X1], S[X2], S.e_roa, fit40)),
          title="Параметрическая модель: плоскость").show()
```

### Взаимодействия: как плоскость перестаёт быть плоской

Пока в модели только слагаемые, поверхность обязана быть плоскостью. Это сильное
утверждение: **эффект $X_1$ один и тот же при любом $X_2$**.

Здесь оно заведомо неверно, и понять это можно не глядя в данные. Списание
кредита бьёт по прибыли пропорционально тому, **какую долю баланса занимают
кредиты**. У банка, вложившего в кредиты все депозиты, кредитный портфель — это
почти весь работающий актив; у банка, держащего половину в ценных бумагах,
кредиты — половина баланса, и тот же процент списаний стоит вдвое дешевле.

Значит, эффект списаний должен зависеть от отношения кредитов к депозитам.
Проверяем.

```python
inter = smf.ols("e_roa ~ " + " + ".join(FEATURES) + f" + {X1}:{X2}",
                data=model_df).fit()
print(inter.summary().tables[1].as_text()[-620:])
print(f"\nR^2:  без взаимодействия {linear.rsquared:.4f}   ->   с ним {inter.rsquared:.4f}")
```

Коэффициент при произведении отрицателен, $t = -44$. Гипотеза подтвердилась.

А вот $R^2$ вырос всего на 0.009 — и это **нормально**. Взаимодействие почти
никогда не двигает $R^2$ заметно: оно не добавляет новой информации, а
**перераспределяет уже имеющуюся** между наблюдениями. Мерить его пользу приростом
$R^2$ — всё равно что мерить пользу очков средней остротой зрения по стране.
Мерить надо **предельным эффектом**.

```python
b1 = inter.params[X1]
bx = inter.params[f"{X1}:{X2}"]
print("сколько ROA стоит рост списаний на 1 процентный пункт:")
for q in (0.1, 0.5, 0.9):
    v = model_df[X2].quantile(q)
    print(f"   кредиты/депозиты = {v:.2f} (перцентиль {int(q*100):2d}) :  "
          f"{(b1 + bx * v) * 0.01 * 1e4:+6.1f} базисных пунктов")
print(f"\nмодель без взаимодействия: всегда "
      f"{linear.params[X1] * 0.01 * 1e4:+.1f} б.п., при любой структуре баланса")
```

Вот теперь видно, зачем это нужно. Один и тот же процентный пункт списаний
стоит банку, вложившему в кредиты половину депозитов, **104 базисных пункта**
ROA, а банку, вложившему всё, — **199**. Разница почти вдвое. Модель без
взаимодействия обещает всем одинаковые **154** — то есть систематически
недооценивает риск у самых закредитованных банков и переоценивает у осторожных.

Прирост $R^2$ был 0.009. Ошибка в оценке риска — двукратная. Вот почему $R^2$ не
тот инструмент.

### Как выглядит «изгибание пространства»

Три поверхности подряд, на одних осях и в одной шкале. Отличается только то,
что модели **разрешено** делать с формой.

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

knn = make_pipeline(StandardScaler(), KNeighborsRegressor(n_neighbors=200))
knn.fit(model_df[[X1, X2]], model_df.e_roa)
Z_knn = knn.predict(pd.DataFrame({X1: XG.ravel(), X2: YG.ravel()})).reshape(XG.shape)
SURF = [("1. линейная: плоскость", partial_surface(linear)),
        ("2. + взаимодействие: скручена", partial_surface(inter)),
        ("3. без формы вовсе: как легли данные", Z_knn)]
zlo = min(Z.min() for _, Z in SURF); zhi = max(Z.max() for _, Z in SURF)
print("диапазон ROA на поверхностях: %.3f .. %.3f" % (zlo, zhi))
```

Смотрите на форму: плоская → скрученная → произвольная.

```python
fig = make_subplots(rows=1, cols=3, specs=[[{"type": "surface"}] * 3],
                    subplot_titles=[t for t, _ in SURF])
for j, (_, Z) in enumerate(SURF, start=1):
    fig.add_surface(z=Z, colorscale="YlGnBu_r", showscale=False, opacity=0.9,
                    row=1, col=j)
def ax(g, n):
    i = np.linspace(0, 29, 4)
    return dict(title=n, tickvals=i, ticktext=[f"{g[int(round(k))]:.2g}" for k in i])
sc = dict(xaxis=ax(xg, "списания"), yaxis=ax(yg, "кредиты/деп"),
          zaxis=dict(title="ROA", range=[zlo, zhi]),
          camera=dict(eye=dict(x=1.9, y=1.0, z=0.7)))
fig.update_layout(height=440, margin=dict(l=0, r=0, t=60, b=0),
                  scene=sc, scene2=sc, scene3=sc)
fig.show()
```

1. **Плоскость.** Наклон по списаниям один и тот же в любой точке. Модель
   утверждает, что структура баланса на цену списаний не влияет.
2. **Скрученная.** Всё ещё «линейчатая» — любой её срез остаётся прямой, — но
   прямые разного наклона. Дальний угол (много списаний **и** много кредитов)
   провалился заметно глубже, чем на плоскости. Это и есть «эффект $X_1$ зависит
   от $X_2$».
3. **Произвольная.** Здесь формы не предполагали вообще: поверхность построена
   методом ближайших соседей — значение в точке равно среднему ROA двухсот самых
   похожих банков. Никакой формулы за ней нет, есть только данные.

Третья картинка показывает то, чего первые две по построению увидеть не могли:
провал не гладкий, у него есть край. Заодно видно, что при больших списаниях
поверхность становится «рваной» — там банков мало, и каждая неровность
нарисована буквально десятком наблюдений.

### Непараметрические модели и цена гибкости

Третья поверхность — пример **непараметрической** модели. Она ничего не
предполагает о форме $f$ и потому может поймать что угодно. Ручка гибкости у неё
одна: сколько соседей усреднять. Покрутим её от края до края.

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

tr, te = train_test_split(model_df, train_size=3000, random_state=SEED)
rows = []
for k in (1, 5, 50, 200, 1000):
    m = make_pipeline(StandardScaler(), KNeighborsRegressor(n_neighbors=k))
    m.fit(tr[[X1, X2]], tr.e_roa)
    rows.append({"соседей k": k,
                 "MSE обучение": mean_squared_error(tr.e_roa, m.predict(tr[[X1, X2]])) * 1e6,
                 "MSE тест": mean_squared_error(te.e_roa, m.predict(te[[X1, X2]])) * 1e6})
pd.DataFrame(rows).set_index("соседей k").round(2)
```

Прочитайте эту таблицу внимательно — в ней весь §7 в миниатюре.

* **$k = 1$: ошибка на обучении ровно ноль.** Каждый банк сам себе ближайший
  сосед, поэтому модель воспроизводит обучающую выборку идеально. И при этом
  на новых данных она **худшая из всех** — в 1.7 раза хуже, чем усреднение по
  пятидесяти соседям. Мы аккуратно выучили и сигнал, и $\varepsilon$.
* **$k = 1000$: другая крайность.** Усредняем такую широкую окрестность, что от
  различий между банками ничего не осталось; ошибка растёт с обоих концов.
* **Минимум посередине**, при $k \approx 50$.

Это **переобучение** и **недообучение** на одной таблице, на настоящих данных.
Именно поэтому «идеально село на данные» — не комплимент модели.

| | Параметрические | Непараметрические |
|---|---|---|
| Что предполагают | форму $f$ | почти ничего |
| Что оценивают | несколько чисел | саму поверхность |
| Сколько данных нужно | мало | много |
| Главный риск | форма угадана неверно | переобучение |
| Что можно рассказать | коэффициенты словами | «вот картинка» |

## 4 · Гибкость против интерпретируемости

Если непараметрические модели гибче, зачем нужны негибкие? Две причины, и вторая
неочевидна.

**Причина первая — интерпретируемость.** Плоскость из §3 умещается в одну фразу:
«процентный пункт капитала стоит 5.7 базисных пункта ROA». Сплайн не умещается
никуда. Если задача — *вывод*, гибкая модель бесполезна, какой бы точной ни была.

В банковском деле это не вопрос вкуса. Модель, по которой отказывают в кредите,
должна уметь объяснить отказ — этого требует регулятор. Поэтому банки до сих пор
деплоят логистическую регрессию там, где бустинг точнее.

**Причина вторая — точность.** Гибкая модель нередко предсказывает **хуже**
жёсткой. Звучит парадоксально; §7 покажет механизм.

Разложим методы курса по этим двум осям.

```python
methods = pd.DataFrame([
    ("Подмножества признаков, лассо", 1.0, 5.0, "S03"),
    ("Линейная регрессия / МНК",      2.0, 4.8, "S02"),
    ("Логистическая регрессия",       2.4, 4.4, "S04"),
    ("Обобщённые аддитивные модели",  3.3, 3.2, "S02"),
    ("Деревья решений",               3.6, 3.6, "S05"),
    ("Бустинг, случайный лес",        4.4, 1.7, "S05"),
    ("Нейросети",                     4.9, 1.0, "S05"),
], columns=["method", "flex", "interp", "session"])

fig = px.scatter(methods, x="flex", y="interp", text="method", color="session",
                 color_discrete_sequence=[BLUE, ORANGE, GREEN, "#CC79A7", "#E69F00"])
fig.update_traces(textposition="top center", marker=dict(size=13))
fig.update_layout(title="Карта методов: за гибкость платят интерпретируемостью",
                  xaxis_title="Гибкость →", yaxis_title="Интерпретируемость →",
                  xaxis=dict(range=[0.4, 5.6], showticklabels=False),
                  yaxis=dict(range=[0.3, 5.9], showticklabels=False),
                  legend_title="сессия", **LAYOUT)
fig.show()
```

Курс идёт по этой карте слева направо: S02–S03 — левый верхний угол, S05 —
правый нижний. На каждом шаге будем спрашивать, стоила ли добавленная гибкость
потерянной понятности.

Вопрос «какую модель взять» **не** решается фразой «возьмём самую современную».
Он решается вопросом «что мне нужно на выходе — число или объяснение?».

## 5 · С учителем и без учителя

Всё, что было выше, — **обучение с учителем** (supervised): для каждого банка
известен правильный ответ $y_i$ (его ROA). Учитель — это $y$.

**Обучение без учителя** (unsupervised): есть только $X$, отклика нет вообще.
Спросить «насколько точно?» не у кого — правильного ответа не существует.
Вопрос меняется на «есть ли здесь структура и какая?».

Банковский вопрос ровно такой: **какие бывают бизнес-модели банков?** Никто не
выдал нам ярлык «этот банк розничный, а этот работает на рынке ценных бумаг» —
но по отчётности это должно быть видно. Возьмём срез на конец 2023 года.

```python
FEATS = ["c_equity_assets", "a_npl_gross_loans", "m_cost_income", "l_loans_deposits",
         "s_securities_assets", "l_brokered_share", "e_nim", "a_cre_concentration"]
snap23 = banks[(banks.repdte == "2023-12-31")].dropna(subset=FEATS).copy()
print("банков в срезе 2023Q4:", len(snap23))
snap23[FEATS].describe().T[["mean", "50%", "std"]].round(3)
```

Восемь измерений глазами не увидеть. **Метод главных компонент** (PCA) сжимает
их в два новых признака так, чтобы сохранить как можно больше различий между
банками. Затем **k-means** разложит банки на три группы, не зная о них ничего.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

Z = StandardScaler().fit_transform(snap23[FEATS])   # общий масштаб: доли и разы несравнимы
pca = PCA(n_components=2).fit(Z)
P = pca.transform(Z)
labels = KMeans(n_clusters=3, n_init=10, random_state=SEED).fit_predict(P)

print("доля сохранённых различий:", pca.explained_variance_ratio_[:2].round(3),
      "= %.1f%%" % (100 * pca.explained_variance_ratio_[:2].sum()))
print("\nиз чего сделана 1-я компонента:")
print(pd.Series(pca.components_[0], index=FEATS).round(2).sort_values().to_string())
```

Две компоненты удержали лишь 43% различий — заметно меньше, чем хотелось бы.
Это честный результат: настоящая банковская отчётность многомернее, чем две оси,
и сводить её к плоскости — уже потеря. (В учебниках PCA обычно показывают на
данных, где двух компонент хватает на 90%. Реальность выглядит иначе.)

Зато первая компонента осмысленна: на одном её конце — `l_loans_deposits`
(кредиты к депозитам), на другом — `s_securities_assets` (ценные бумаги).
Это ось **«кредитный банк против банка ценных бумаг»** — реальное различие
бизнес-моделей, а не артефакт.

```python
viz = pd.DataFrame({
    "PC1": P[:, 0], "PC2": P[:, 1],
    "кластер": [f"кластер {l}" for l in labels],
    "банк": snap23["name"].to_numpy(), "штат": snap23["state"].to_numpy(),
    "активы, млн $": (snap23["assets"] / 1000).round(0).to_numpy(),
    "кредиты/депозиты": snap23["l_loans_deposits"].round(2).to_numpy(),
    "ценные бумаги/активы": snap23["s_securities_assets"].round(2).to_numpy()})
fig = px.scatter(viz, x="PC1", y="PC2", color="кластер",
                 hover_data=["банк", "штат", "активы, млн $",
                             "кредиты/депозиты", "ценные бумаги/активы"],
                 color_discrete_sequence=[BLUE, ORANGE, GREEN])
fig.update_traces(marker=dict(size=7, opacity=0.75))
fig.update_layout(title="Банки в двух главных компонентах (наведите курсор на точку)",
                  xaxis_title="PC1: ценные бумаги  ←→  кредиты",
                  yaxis_title="PC2: качество активов", **LAYOUT)
fig.show()
```

Группы разделились. Но **правильный ли это ответ?** Спросить не у кого — в этом
и суть обучения без учителя. Единственное, что можно сделать, — посмотреть, чем
кластеры содержательно различаются, и попробовать их **назвать**.

```python
prof = snap23.groupby(labels)[FEATS].median().round(3)
prof.insert(0, "банков", pd.Series(labels).value_counts().sort_index().to_numpy())
prof["активы, млн $"] = (snap23.groupby(labels)["assets"].median() / 1000).round(0)
prof[["банков", "l_loans_deposits", "s_securities_assets", "a_npl_gross_loans",
      "l_brokered_share", "m_cost_income", "активы, млн $"]]
```

Названия напрашиваются сами:

* **кластер 0 — казначейский** (213 банков): треть активов в ценных бумагах,
  кредитов лишь 63 копейки на рубль депозитов. Такой банк зарабатывает на
  портфеле, а не на кредитовании.
* **кластер 1 — кредитный** (304 банка): кредиты почти равны депозитам (0.89),
  бумаг вдвое меньше. Классический «депозиты собрал — кредиты выдал».
* **кластер 2 — проблемный** (15 банков): просрочка 5.5% против 0.5–0.6% у
  остальных, 16% фондирования брокерское (дорогое, убегает первым), издержки
  съедают доход целиком (1.44). Пятнадцать банков из пятисот, и по отчётности
  видно, что у них что-то не так.

> **Честная оговорка.** Мы не *проверили* эту таксономию — мы её *предложили*.
> Метрики качества, аналогичной MSE, здесь нет. Результат обучения без учителя
> аргументируют, а не измеряют: называют кластеры содержательно и проверяют на
> устойчивость (переживут ли они другой квартал, другую выборку, другое $k$).
> Этим займёмся в S06, где та же задача — бизнес-модели банков — будет
> центральной.

## 6 · Регрессия и классификация

Ещё одно деление, и зависит оно **только от типа отклика $Y$**.

* $Y$ **количественный** (число: ROA, активы, маржа) → **регрессия**
* $Y$ **качественный** (категория: упал / не упал, класс лицензии) → **классификация**

Признаки $X$ могут быть любыми в обоих случаях — тип $X$ на это деление не влияет.

До сих пор мы решали задачу регрессии. Теперь главный вопрос курса — **упадёт ли
банк в ближайший год?** Отклик качественный: да или нет.

```python
print("распределение отклика failed_next_4q:")
print(banks.failed_next_4q.value_counts().to_string())
print(f"\nдоля крахов: {banks.failed_next_4q.mean():.2%}")
print("\nмедианы по группам:")
print(banks.groupby("failed_next_4q")[
    ["c_equity_assets", "a_npl_gross_loans", "e_roa", "m_cost_income"]].median().round(4).to_string())
```

Различия огромные: перед крахом капитал в два с половиной раза тоньше (3.9%
против 9.8%), просрочка в одиннадцать раз выше (11.1% против 1.0%), прибыль
отрицательная, а издержки превышают процентный доход. Сигнал в данных есть, и
сильный.

Почему для такого $Y$ не годится обычный МНК? Потому что квадрат ошибки
$(Y - \hat Y)^2$ требует вычитания, а `"да" − "нет"` не определено. Закодировать
классы числами 0/1 и минимизировать MSE технически можно — такая модель
называется **линейной вероятностной**. Посмотрим, что она выдаёт.

```python
import statsmodels.api as sm

d = banks.copy()
d["y"] = d.failed_next_4q.astype(int)
lpm = smf.ols("y ~ a_npl_gross_loans", data=d).fit()
logit = smf.glm("y ~ a_npl_gross_loans", data=d, family=sm.families.Binomial()).fit()
g = pd.DataFrame({"a_npl_gross_loans": np.linspace(0, 0.35, 120)})
print(f"ЛВМ при нулевой просрочке предсказывает вероятность краха "
      f"{lpm.predict(pd.DataFrame({'a_npl_gross_loans': [0.0]})).iloc[0]:+.4f}")
print(f"логит — {logit.predict(pd.DataFrame({'a_npl_gross_loans': [0.0]})).iloc[0]:.4f}")
print(f"\nлогит: +1 п.п. просрочки умножает ШАНСЫ краха на "
      f"{np.exp(logit.params['a_npl_gross_loans'] * 0.01):.2f}")
```

Линейная вероятностная модель предсказывает вероятность краха **−1.9%** для
здорового банка. Это не «неточность» — это бессмыслица, и на графике она видна.

```python
obs = (d.assign(bin=pd.cut(d.a_npl_gross_loans, np.arange(0, 0.36, 0.02)))
         .groupby("bin", observed=True)
         .agg(npl=("a_npl_gross_loans", "mean"), rate=("y", "mean"), n=("y", "size")))
fig = go.Figure()
fig.add_hrect(y0=-0.1, y1=0, fillcolor=ORANGE, opacity=0.08, line_width=0)
fig.add_scatter(x=g.a_npl_gross_loans, y=lpm.predict(g), mode="lines",
                name="линейная вероятностная модель (МНК)", line=dict(color=ORANGE, width=3))
fig.add_scatter(x=g.a_npl_gross_loans, y=logit.predict(g), mode="lines",
                name="логистическая регрессия", line=dict(color=BLUE, width=3))
fig.add_scatter(x=obs.npl, y=obs.rate, mode="markers", name="наблюдаемая доля крахов",
                marker=dict(color="#333333", size=9))
fig.add_hline(y=0, line=dict(color=GREY, width=1))
fig.add_hline(y=1, line=dict(color=GREY, width=1))
fig.update_layout(title="Вероятность не бывает отрицательной",
                  xaxis_title="Проблемные кредиты / кредиты",
                  yaxis_title="P(крах в ближайшие 4 квартала)",
                  yaxis=dict(range=[-0.1, 1.05]), **LAYOUT)
fig.show()
```

Логит по построению живёт в $(0, 1)$ и воспроизводит S-образную форму, которую
видно в наблюдаемых долях: до 5% просрочки почти ничего не происходит, между 10%
и 25% риск взлетает, дальше выходит на плато.

Ценой становится чтение коэффициентов: они теперь в логарифме шансов, а
предельный эффект зависит от точки. Полностью — в S04.

> **Как мерить качество классификатора — отдельный разговор.** Заметьте: крахов
> здесь всего полтора процента. Что это делает с привычными метриками и какой
> порог выбирать, если ошибки стоят по-разному, — тема S04. Пока достаточно
> понимать, **почему** это другая задача, а не как её оценивать.

> **Граница размытее, чем кажется.** Логистическая регрессия выдаёт **число** —
> вероятность, — и лишь потом мы сами превращаем его в класс, сравнивая с
> порогом. Вопрос «регрессия или классификация» — это чаще про то, какое решение
> принимается на выходе, чем про сам метод.

## 7 · Как измерить точность

Для регрессии стандартная мера — **среднеквадратичная ошибка**:

$$ \mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}\big(y_i - \hat f(x_i)\big)^2 $$

Вопрос в том, **на каких** $y_i$ её считать. Если на тех же, на которых модель
училась, — это **обучающая** MSE, и она почти бесполезна: её всегда можно
загнать в ноль, мы это видели на сплайне в §3. Нас интересует **тестовая** MSE —
на данных, которых модель не видела.

Проверим: предскажем ROA по доле проблемных кредитов. Учимся на 1500 банк-кварталах,
проверяем на остальных ста тысячах, и повышаем гибкость — степень полинома.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, MinMaxScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

dz = banks.dropna(subset=["e_roa", "a_npl_gross_loans"])
Xtr, Xte, ytr, yte = train_test_split(dz[["a_npl_gross_loans"]], dz.e_roa,
                                      train_size=1500, random_state=7)
rows, models = [], {}
for deg in range(1, 9):
    pipe = make_pipeline(MinMaxScaler((-1, 1)), PolynomialFeatures(deg), LinearRegression())
    pipe.fit(Xtr, ytr)
    models[deg] = pipe
    rows.append({"степень": deg, "степеней свободы": deg + 1,
                 "MSE обучение": mean_squared_error(ytr, pipe.predict(Xtr)) * 1e6,
                 "MSE тест": mean_squared_error(yte, pipe.predict(Xte)) * 1e6})
curve = pd.DataFrame(rows).set_index("степень")
print(f"обучение: {len(Xtr):,} строк | тест: {len(Xte):,} строк   (MSE умножена на 10^6)")
curve.round(2)
```

**Степени свободы** — число подгоняемых чисел. У полинома степени $d$ их $d+1$.
Чем больше, тем гибче модель.

Обучающая MSE падает **монотонно** — она обязана падать. Тестовая ведёт себя
иначе.

```python
fig = go.Figure()
fig.add_scatter(x=curve.index, y=curve["MSE обучение"], mode="lines+markers",
                name="обучающая MSE", line=dict(color=BLUE, width=3))
fig.add_scatter(x=curve.index, y=curve["MSE тест"], mode="lines+markers",
                name="тестовая MSE", line=dict(color=ORANGE, width=3, dash="dash"))
best = curve["MSE тест"].idxmin()
fig.add_vline(x=best, line=dict(color=GREY, dash="dot"),
              annotation_text=f"минимум теста: степень {best}")
fig.update_layout(title="U-образная кривая: гибкость помогает, потом вредит",
                  xaxis_title="Степень полинома (гибкость →)",
                  yaxis_title="MSE × 10⁶ (лог. шкала)", yaxis_type="log", **LAYOUT)
fig.show()
```

Синяя линия ползёт вниз всё время — она обязана: чем гибче модель, тем ближе
она к своим точкам. Оранжевая опускается ровно на один шаг, а дальше растёт — и
к восьмой степени оказывается **в шестьсот тысяч раз** хуже, чем в минимуме.

Обратите внимание, каким **неглубоким** получился этот минимум: между степенью 1
и степенью 2 разница меньше процента. Немного кривизны помогает чуть-чуть; много
гибкости уничтожает модель. Асимметрия чудовищная — и на реальных данных так
бывает чаще, чем в учебниках.

Разрыв между синей и оранжевой линиями и называется **переобучением**.
Посмотрим на три подгонки глазами.

```python
grid = pd.DataFrame({"a_npl_gross_loans": np.linspace(0, 0.30, 300)})
fig = go.Figure()
fig.add_scatter(x=Xtr.a_npl_gross_loans, y=ytr, mode="markers", name="обучающие банки",
                marker=dict(color=GREY, size=4, opacity=0.5))
for deg, col, nm in [(1, BLUE, "степень 1: прямая"),
                     (int(curve['MSE тест'].idxmin()), GREEN, "лучшая по тесту"),
                     (8, ORANGE, "степень 8: заучила шум")]:
    fig.add_scatter(x=grid.a_npl_gross_loans, y=models[deg].predict(grid), mode="lines",
                    name=nm, line=dict(color=col, width=3))
fig.update_layout(title="Одни и те же данные, три степени гибкости",
                  xaxis_title="Проблемные кредиты / кредиты", yaxis_title="ROA",
                  yaxis=dict(range=[-0.10, 0.05]), **LAYOUT)
fig.show()
```

Прямая (синяя) не успевает за изгибом облака. Зелёная ложится на данные.
Полином 8-й степени (оранжевый) виляет между точками и на краях, где банков
почти нет, уходит в никуда.

### Смещение и дисперсия

У этих двух способов ошибиться есть имена, и ожидаемая ошибка точно
раскладывается на них:

$$
\mathbb{E}\big[(y_0 - \hat f(x_0))^2\big]
= \underbrace{\operatorname{Var}(\hat f(x_0))}_{\text{дисперсия}}
+ \underbrace{\big[\operatorname{Bias}(\hat f(x_0))\big]^2}_{\text{смещение}^2}
+ \underbrace{\operatorname{Var}(\varepsilon)}_{\text{неснижаемая}}
$$

* **Смещение** — ошибка от того, что модель слишком проста для истинной $f$.
  У прямой на картинке выше смещение большое.
* **Дисперсия** — насколько сильно $\hat f$ дёргалась бы, дай мы модели другую
  обучающую выборку. У полинома 10-й степени дисперсия огромная.
* **Неснижаемая** — знакомый пол из §2.

Обе первых величины неотрицательны, значит тестовая MSE **никогда не опускается
ниже** $\operatorname{Var}(\varepsilon)$.

Увидеть их по отдельности на реальных данных невозможно: нужна истинная $f$ и
много обучающих выборок. Поэтому — симуляция.

```python
N_REP, N_TRAIN = 200, 40
xg = np.linspace(-3, 3, 60); fg = f_true(xg)
preds = {d: np.empty((N_REP, xg.size)) for d in range(1, 11)}
for r in range(N_REP):
    xt = rng.uniform(-3, 3, N_TRAIN); yt = f_true(xt) + rng.normal(0, sigma, N_TRAIN)
    for d in preds:
        preds[d][r] = np.polyval(np.polyfit(xt, yt, d), xg)

dec = pd.DataFrame({
    "смещение²": [np.mean((preds[d].mean(0) - fg) ** 2) for d in preds],
    "дисперсия": [np.mean(preds[d].var(0)) for d in preds]}, index=list(preds))
dec["неснижаемая"] = sigma ** 2
dec["сумма трёх"] = dec.sum(axis=1)
dec.round(3)
```

Слева и в центре — по 30 подгонок каждой степени на разных выборках. Справа —
разложение суммой.

```python
fig = make_subplots(rows=1, cols=3, subplot_titles=[
    "степень 1: кучно, но мимо", "степень 10: в среднем верно, но вразнос",
    "разложение ошибки"])
for j, d in enumerate([1, 10], start=1):
    for r in range(0, N_REP, 7):
        fig.add_scatter(x=xg, y=preds[d][r], mode="lines", row=1, col=j,
                        line=dict(color=BLUE, width=1), opacity=0.22, showlegend=False)
    fig.add_scatter(x=xg, y=fg, mode="lines", row=1, col=j, showlegend=False,
                    line=dict(color="black", width=3, dash="dash"))
    fig.update_yaxes(range=[-4, 4], row=1, col=j)
for nm, col in [("смещение²", BLUE), ("дисперсия", ORANGE), ("сумма трёх", GREEN)]:
    fig.add_scatter(x=dec.index, y=dec[nm], mode="lines+markers", name=nm,
                    line=dict(color=col, width=3), row=1, col=3)
fig.update_yaxes(type="log", row=1, col=3)
fig.update_layout(title="Чёрный пунктир — истинная f, синие линии — подгонки",
                  height=430, margin=dict(l=50, r=20, t=70, b=50))
fig.show()
```

Слева: прямые лежат друг на друге — **дисперсия маленькая**, — но все дружно
мимо: **смещение большое**. В центре: полиномы 10-й степени в среднем проходят
где надо (**смещение почти ноль**), но каждый отдельный виляет по-своему
(**дисперсия огромная**). Справа одно падает, другое растёт, а сумма имеет
минимум посередине.

> **Компромисс смещения и дисперсии.** Хорошая модель — не самая гибкая и не
> самая простая, а та, у которой **сумма** этих двух ошибок минимальна. Найти
> этот минимум по данным, не подглядывая в тест, — задача S03.

## 8 · Итог

1. **$Y = f(X) + \varepsilon$.** Мы оцениваем $f$; $\varepsilon$ — то, чего наши
   признаки не знают.
2. **Предсказание и вывод — разные задачи.** Покрытие резервами значимо в парной
   регрессии ($t = 63$) и пусто в множественной ($p = 0.29$), потому что было
   лишь отражением качества активов.
3. **Неснижаемая ошибка** ставит пол точности. Он неизвестен, но он есть.
4. **Настоящие данные надо чистить руками и вслух.** Мы отбросили треть строк —
   и тем самым сменили популяцию, о которой делаем выводы.
5. **Параметрические модели** сводят задачу к нескольким числам и рискуют
   ошибиться формой; **непараметрические** ничего не предполагают и рискуют
   переобучиться.
6. **Взаимодействия и степени** изгибают поверхность отклика — это и означает
   «эффект $X_1$ зависит от $X_2$». Урон от просрочки для банка с тонким
   капиталом в 1.7 раза выше.
7. **Гибкость покупается интерпретируемостью**, а в банке — ещё и требованиями
   регулятора к объяснимости.
8. **Без учителя** нет правильного ответа: результат аргументируют, а не
   измеряют.
9. **Регрессия или классификация** зависит только от типа $Y$.
10. **Обучающая MSE всегда падает, тестовая — U-образна.** Минимум лежит на
    средней гибкости, потому что ошибка = смещение² + дисперсия + шум.
